```python
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "experiments/brats23_axial_fold1_orig_24551546/epoch_metrics.csv"

df = pd.read_csv(csv_path)

plt.figure(figsize=(8, 5))
plt.plot(df["epoch"], df["dice_mean"], marker="o")
plt.xlabel("Epoch")
plt.ylabel("Best Dice Mean")
plt.title("Best Dice Mean per Epoch")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(df["epoch"], df["train_loss"], marker="o")
plt.xlabel("Epoch")
plt.ylabel("Train Loss")
plt.title("Train Loss per Epoch")
plt.grid(True)
plt.tight_layout()
plt.show()
```
